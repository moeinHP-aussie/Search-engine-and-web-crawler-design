### ۴.۱ Mapping سفارشی

در Elasticsearch، مفهوم Mapping فرآیند تعیین ساختار سند و نحوه‌ی برخورد موتور جستجو با فیلدهای مختلف است. مپینگ صراحتاً مشخص می‌کند که هر فیلد از چه نوع داده‌ای (Data Type) است و فرآیند آنالیز و توکن‌سازی روی آن به چه شکلی انجام شود تا برای جستجوهای آینده آماده گردد.

```python
INDEX_MAPPING = {
    "settings": {
        "number_of_shards": 1,
        "number_of_replicas": 0,
        "analysis": {
            "analyzer": {
                "persian_analyzer": {
                    "tokenizer": "standard",
                    "filter": ["lowercase", "persian_stop_filter"]
                }
            }
        }
    },
    "mappings": { ... }
}
```

آنالایزر سفارشی `persian_analyzer` دو فیلتر دارد:
- **lowercase:** تبدیل به حروف کوچک (برای انگلیسی)
- **persian_stop_filter:** حذف ۳۳ stop word فارسی پرتکرار

### ۵.۱ الگوریتم BM25
Elasticsearch به‌صورت پیش‌فرض از الگوریتم **BM25 (Best Match 25)** برای رتبه‌بندی اسناد استفاده می‌کند. BM25 یک فرمول پیشرفته مبتنی بر TF-IDF است که دو بهبود ساختاری مهم (اشباع فرکانس ترم و نرمال‌سازی طول سند) دارد.

**فرمول ریاضی BM25:**

$$\text{score}(q, d) = \sum_{t \in q} \text{IDF}(t) \cdot \frac{tf(t,d) \cdot (k_1 + 1)}{tf(t,d) + k_1 \cdot \left(1 - b + b \cdot \frac{|d|}{\text{avgdl}}\right)}$$

| پارامتر | توضیح | مقدار پیش‌فرض |
| :--- | :--- | :---: |
| **$tf(t,d)$** | تعداد تکرار کلمه (term) در سند جاری | — |
| **$\text{IDF}(t)$** | معکوس فراوانی کلمه در کل مجموعه اسناد (Corpus) | — |
| **$k_1$** | کنترل سرعت اشباع ترم (جلوگیری از تأثیر بی‌نهایت فرکانس کلمه) | `1.2` |
| **$b$** | میزان اثرگذاری جریمه‌ی طول سند (نرمال‌سازی طول اسناد) | `0.75` |
| **$|d|$** | طول سند جاری (تعداد کلمات سند) | — |
| **$\text{avgdl}$** | میانگین طول تمام اسناد موجود در ایندکس | — |

### ۵.۲ وزن‌دهی فیلدها (Field Boosting)
وزن‌دهی به فیلدها با نماد `^` در پارامتر `fields` اعمال می‌شود تا ارزش فیلدهای حیاتی‌تر (مانند عنوان) در رتبه‌بندی نهایی بیشتر شود.

**فرمول محاسباتی امتیاز نهایی با مکانیزم `best_fields`:**

$$\text{final\_score} = \max(\text{Score of Best Field}) + \left( \text{tie\_breaker} \times \sum \text{Scores of Other Fields} \right)$$

در این ساختار، امتیاز هر سند ابتدا در فیلدهای مختلف محاسبه شده و با ضرایب زیر اعمال می‌شود:
* **وزن بالا ($\times 3.0$):** فیلدهای `title` و `title_tokens`
* **وزن متوسط ($\times 2.0$):** فیلدهای `headings` و `headings_tokens`
* **وزن پایه ($\times 1.0$):** فیلدهای `body` و `body_tokens`

---

### ۵.۳ ساختار Query اصلی

```python
query_body = {
    "query": {
        "multi_match": {
            "query":       query_text,
            "fields":      ["title^3", "title_tokens^3",
                            "headings^2", "headings_tokens^2",
                            "body^1", "body_tokens^1"],
            "type":        "best_fields",
            "tie_breaker": 0.3,
            "fuzziness":   "AUTO",   # جستجوی تقریبی برای اشتباهات تایپی
        }
    },
    "size": k,
    "highlight": { ... }
}

**`tie_breaker=0.3`** یعنی: ۳۰٪ امتیاز فیلدهای دیگر (غیر از بهترین) هم به امتیاز نهایی اضافه می‌شود.
